### 🧠 منطق مسیریابی

**Attachment**  
نشانه‌های واضح دلبستگی اضطرابی دیده می‌شود: ترس از رها شدن، نیاز به اطمینان گرفتن مداوم، احساس «هرگز در این رابطه احساس امنیت نکردم»، و نوسان بین میل به نزدیکی و اجتناب.

**Relational Pattern**  
الگوی رابطه‌ای سمی توصیف شده است: چرخه دعوا–خوشحالی–دعوا، تحقیر، دستکاری روانی، ایجاد احساس گناه، نقض مرزها و پویایی‌های قدرت.

**Cognitive Distortion**  
تحریف‌های شناختی مانند تفکر سیاه‌وسفید («آیا من آدمی بدون تعهد اخلاقی هستم؟»)، فاجعه‌سازی، سرزنش شدید خود و نشخوار ذهنی مداوم («این همه چرا در ذهنم»).

**Schema**  
باورهای هسته‌ای مربوط به بی‌ارزشی دیده می‌شود («چرا او نمی‌خواهد من احساس دوست‌داشتنی و باارزش بودن داشته باشم؟») همراه با طرحواره نقص/شرم و نیاز به تأیید دیگران در الگوهای رابطه‌ای.

## **تحلیل سبک دلبستگی**

### **نمره‌های سبک دلبستگی (از ۰ تا ۱۰)**

| سبک دلبستگی | نمره |
|:---|:---:|
| دلبستگی ایمن | ۱ |
| دلبستگی دلواپس‑مشغول | ۹ |
| دلبستگی دوری‌گزین‑طردکننده | ۴ |
| دلبستگی آشفته | ۷ |

### **خلاصه تحلیل**

فرد نشانه‌های شدید سبک **دلبستگی دلواپس‑مشغول** را بروز می‌دهد: ترس از طرد شدن، نشخوار فکری مداوم درباره رفتار شریک عاطفی، نیاز مفرط به تأیید و امنیت، و احساس بی‌ارزشی مگر از طریق دیگری.  

همزمان، گرایش‌هایی به سبک **دلبستگی دوری‌گزین‑طردکننده** دیده می‌شود: ابراز نیاز شدید به آزادی و استقلال، نخواستن تعهد، و اجتناب از صمیمیت واقعی.  

الگوی غالب اما **آشفته** است: همزمان هم خواهان رابطه است و هم از آن می‌گریزد، در دوگانگی شدید به سر می‌برد، احساس خفگی و به‌دام‌افتادگی می‌کند، و رابطه را چرخه‌ای تکراری از قهر، دعوا، خوشحالی عمیق و سپس فروپاشی توصیف می‌کند.  

شواهدی از **دلبستگی ایمن** وجود ندارد. فرد آگاهانه به ناامنی مزمن، شنیده نشدن، و نقش اثبات‌کردن مداوم خود در رابطه اشاره دارد.

## الگوهای رابطه‌ای

### نمره‌ها (از ۰ تا ۱۰)

| الگوی رابطه‌ای | نمره |
|:---|:---:|

| چرخه پس‌زدن‑چسبیدن | ۹ |
| رفتار کنترل‌گرانه | ۷ |
| حسادت بیمارگون | ۵ |
| خیانت مکرر | ۱ |

### خلاصه

فرد چرخه تکراری از دعوا، جدایی و آشتی را توصیف می‌کند (الگوی تعارض مکرر ۹) با فرازونشیب‌های عاطفی شدید. وابستگی عاطفی شدیدی به شریک زندگی دارد با وجود اینکه خواهان آزادی است (وابستگی ناسالم ۸). متن به وضوح پویایی پس‌زدن‑چسبیدن را نشان می‌دهد: فرد درخواست قطع ارتباط می‌کند اما امیدوار است سال‌ها بعد طرف مقابل تغییر کند، و هم احساس به دام افتادن دارد و هم نمی‌تواند ترک کند (چرخه پس‌زدن‑چسبیدن ۹). شریک زندگی رفتارهای کنترل‌گرانه نشان می‌دهد مانند اینکه می‌پرسد چرا حلقه را نگه داشته، درباره لباس می‌گوید «نمی‌دونم کی می‌خواد توی تنت ببینه»، و تأکید می‌کند که نباید کسی به بدن فرد دست بزند (رفتار کنترل‌گرانه ۷). حسادت خفیفی وقتی شریک با همکارش لاس می‌زند دیده می‌شود، اما فرد بیشتر احساس ترحم می‌کند تا حسادت (حسادت بیمارگون ۵). شواهدی از خیانت مکرر وجود ندارد (خیانت مکرر ۱).

### خلاصه تحلیل

فرد به‌طور مشخص طرحواره **رها شدگی/بی‌ثباتی** را با نمره ۹ (بسیار شدید) نشان می‌دهد: ترس دائمی از پایان رابطه، بازگشت به چرخه‌های تکراری، و نگرانی از اینکه طرف مقابل تغییر نخواهد کرد. همزمان، **منفی‌گرایی/بدبینی** نیز نمره ۹ دارد: انتظار شکست، تحقیر، و بدترین سناریوها در رابطه.

**بی‌اعتمادی/بدرفتاری** با نمره ۸ (شدید) نمایان است: احساس می‌کند طرف مقابل دنبال تحقیر، توهین و تهمت زدن است، و نمی‌تواند به او اعتماد کند. **نقص/شرم** با نمره ۸: خود را به خاطر یادآوری فرد دیگر در لحظه صمیمیت، شدیداً سرزنش می‌کند و می‌پرسد «مشکل من چیست؟» و «آدم بی‌تعهدی هستم؟».

**جلب تأیید/جلب توجه** با نمره ۸: نیاز دارد طرف مقابل اشتباهش را قبول کند و به او بگوید آدم بدی نیست، و مدام از خود می‌پرسد «چرا نمی‌خواد احساس کنم دوست‌داشتنی و ارزشمندم؟».

**محرومیت هیجانی** با نمره ۷: اذعان می‌کند «هرگز امنیت نداشتم، شنیده نشدم». **درهم‌تنیدگی/خود رشدنیافته** با نمره ۷: حس می‌کند با دیدن طرف مقابل دوباره به ۱۸ سالگی برگشته و هویت خود را گم کرده. **تسلیم** با نمره ۷: احساس خفگی و تعهد اجباری دارد، اما نمی‌خواهد تسلیم شود. **معیارهای سخت‌گیرانه** با نمره ۷: خود را مدام قضاوت می‌کند که کار درست چیست و آیا اشتباه کرده است.

**انزوای اجتماعی/بیگانگی** با نمره ۶: در اتاق خود را حبس می‌کند، از پدر فاصله می‌گیرد، و احساس تنهایی در کشمکش‌های درونی دارد. **خودفداکاری** با نمره ۶: نیازهای خود (آزادی، ثبات) را نادیده گرفته و به رابطه چرخه‌ای بازمی‌گردد. **تنبیه‌گری** با نمره ۶: عذاب وجدان شدید، یک هفته حمام نرفتن به جز روز ملاقات، و سرزنش مداوم خود.

**شکست** با نمره ۶: می‌پرسد «هیچوقت کافی بودم؟» و احساس می‌کند در اثبات خود شکست خورده است. **خودکنترلی ناکافی** با نمره ۵: می‌داند دوباره به همان آدم برمی‌گردد، با وجود آگاهی از الگوی مضر. **مهار هیجانی** با نمره ۵: سعی می‌کند احساسات خود را با تمیز کردن خانه و خواندن کتاب مدیریت کند، اما آشفتگی درونی بالا است.

**وابستگی/نالایقتی** با نمره ۴: احساس گیجی و آشفتگی در تصمیم‌گیری، و تکیه به مادر برای غذا و مراقبت. **آسیب‌پذیری** با نمره ۳: اشاره خفیف به ترس از آسیب جسمی یا روانی. **استحقاق/بزرگ‌منشی** با نمره ۱: تقریباً وجود ندارد.

## 🧠 منطق مسیریابی (توضیحات تحلیلی)

### **Attachment – دلبستگی**
شواهدی از الگوی دلبستگی اضطرابی دیده می‌شود؛ مثل نیاز به تأیید از سمت پدر و شریک عاطفی، احساس نادیده‌گرفته‌ شدن وقتی نیازها برآورده نمی‌شوند، و حس ناامنی در روابط.

### **Cognitive Distortion – تحریف‌های شناختی**
وجود تفکر سیاه‌وسفید («همه عادت دارن»)، تفسیر احساسی، و احتمال فاجعه‌سازی دربارهٔ روابط و ارزشمندی خود.

### **Relational Pattern – الگوهای ارتباطی**
الگوهای پررنگی از تعارض با پدر، اختلال در ارتباط‌گیری، انتظارات برآورده‌نشده در رابطهٔ دوطرفه، و احساس «قدر دانسته نشدن» در روابط.

### **Schema – طرحواره‌ها**
نشانه‌هایی از طرحوارهٔ نقص/شرم («حس کردم خیلی دم‌دستی‌ام»)، محرومیت هیجانی، و رهاشدگی به‌دلیل بی‌توجهی پدر و نیازهای هیجانی برآورده‌نشده.

### خلاصه تحلیل

فرد به‌طور مشخص و بسیار شدید، **محرومیت هیجانی** (نمره ۹) را تجربه می‌کند: «هیچ وقت نتونستم نه ناز کنم نه ناز من کسی بخره»، «کاش وقتی برای بقیه ذوق میکنم اونا هم برای من ذوق کنن»، «وقتی یه متن بلند براشون تایپ میکنم اونا هم همون تایپ کنن». این محرومیت در تمام ابعاد عاطفی (مراقبت، همدلی، حمایت) دیده می‌شود.

**خودفداکاری** (نمره ۸) بسیار برجسته است: «از بس برای همه بودم همه عادت دارن که من باشم»، «ازش عذرخواهی کردم با اینکه میدونستم تقصیر من نبوده». فرد نیازهای خود را فدای دیگران می‌کند و انتظار متقابل دارد اما دریافت نمی‌کند.

**جلب تأیید/جلب توجه** (نمره ۸): «یه چیزایی که منم حس کنم وجود دارم»، «دوست داشتم متین بهم یه پیام طولانی بده»، «دوست داشتم بابام بهم زنگ بزنه یا منو ببره بیرون». ارزش خود را وابسته به توجه و تأیید دیگران می‌بیند.

**تسلیم** (نمره ۷): با وجود اینکه می‌داند تقصیر او نبوده، عذرخواهی می‌کند. احساس می‌کند دیگران هرچیزی بخواهند می‌توانند به او بگویند («چقدر راحت ادما به خودشون اجازه میدن هرچی می خوان بهم بگن»).

**نقص/شرم** (نمره ۷): «هیچ وقت نتونستم رفتار دخترونه داشته باشم»، احساس می‌کند دم‌دستی است و ارزشی ندارد مگر زمانی که برای دیگران مفید باشد.

**رها شدگی/بی‌ثباتی** (نمره ۶): «وقتی قهر میکنم کسی دنبالم نمیاد»، انتظار طرد شدن و رها شدن دارد. همچنین امید به تماس پدر (که اهمیت نمی‌دهد) نشان‌دهنده ترس از بی‌ثباتی در دلبستگی است.

**انزوای اجتماعی/بیگانگی** (نمره ۶): «حتی یه کلمه با مامانم یا کسی حرف نزدم»، احساس تنهایی در روزهای سخت. **مهار هیجانی** (نمره ۶): ابتدای روز را بدون ابراز هیچ احساسی و با سکوت مطلق سپری کرده است. **منفی‌گرایی/بدبینی** (نمره ۶): «با اینکه گذشتم چقدر درد کشیدم»، «بابام بهم اهمیت نمیده».

**درهم‌تنیدگی/خود رشدنیافته** (نمره ۵): هویت فرد گره خورده به انتظارات دیگران و ناتوانی در تمایز نیازهای خود از نیازهای آن‌ها. **شکست** (نمره ۵): احساس می‌کند در «دخترانه بودن» و دیده شدن شکست خورده است. **بی‌اعتمادی/بدرفتاری** (نمره ۵): اشاره به اینکه متین فکر کرده به او دروغ گفته، و پدر مدام مادر و خانواده او را تحقیر می‌کند.

سایر طرحواره‌ها نمرات پایین‌تری دارند. **استحقاق** (نمره ۱) تقریباً وجود ندارد – فرد به جای احساس استحقاق، خود را کم‌ارزش می‌بیند.

**ارزیابی و دسته‌بندی عوامل روانشناختی (ترجمه کامل JSON)**

**عوامل انتخاب‌شده (selected_agents):**
- **attachment (دلبستگی):**  
  شواهدی از دلبستگی اضطرابی به دلیل تحمل طولانی‌مدت بی‌توجهی و بی‌احترامی، و همچنین دلبستگی آشفته (Disorganized) ناشی از اثرات تروما و خیانت.
- **clinical_disorder (اختلال بالینی):**  
  وجود نشانه‌های واضح افسردگی شدید همراه با افکار خودکشی («من دیشب تا حالا مردم»)، ناامیدی عمیق، و احتمال بروز استرس حاد ناشی از ورشکستگی و خیانت.
- **cognetive_distortion (خطاهای شناختی):**

### خلاصه تحلیل

فرد به شدت وابسته به شریک عاطفی خود بوده و با وجود چهارده سال بی‌محبتی، بی‌احترامی، بچه‌ننه بودن، مشکلات مالی شدید، ورشکستگی مکرر، بیکاری، و از دست دادن تمام دارایی، **تحمل کرده و ساخته است** (دلبستگی دلواپس‑مشغول ۹). ارزش خود را کاملاً به بقای رابطه گره زده و با وجود خیانت‌های متعدد، تا لحظه دیدن پیام‌ها زیر بار نرفته است.

اکنون پس از کشف قطعی خیانت، احساس **فروپاشی کامل و «مردن» درونی** می‌کند: «دیگه آدم نمیشم»، «زندگیم ارزش موندن نداره». اما همزمان به عنوان یک مادر، به دنبال راهی برای ادامه دادن (قرص آرامبخش) است (دلبستگی آشفته ۸). الگوی **هم‌زمان خواستن و نخواستن**، **نزدیک شدن و پس زدن**، و **درماندگی همراه با جستجوی راه حل** نمایان است: هم نمی‌تواند بماند و هم نمی‌تواند برود، هم از رابطه متنفر است و هم بدون آن خود را نابود شده می‌بیند.

نشانه‌ای از دلبستگی ایمن وجود ندارد (دلبستگی ایمن ۰). اجتناب خفیف (دلبستگی دوری‌گزین‑طردکننده ۲) فقط در درخواست راه‌حل دارویی برای بی‌حس کردن احساسات و قطع موقت رنج دیده می‌شود، نه در الگوی کلی رابطه. فرد به جای اجتناب فعال از دلبستگی، سال‌ها با چسبیدن به یک رابطه آزاردهنده دوام آورده است.

**نکته مهم بالینی**: فرد افکار خودکشی‌وار («مردم»، «دیگه آدم نمیشم») ابراز می‌کند و نیاز به مداخله فوری حمایتی و درمانی دارد. درخواست قرص آرامبخش نشانه کمک‌خواهی غیرمستقیم است.
